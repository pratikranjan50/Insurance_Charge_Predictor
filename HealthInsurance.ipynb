```python
DATA_PATH = "Health_Insurance.csv"
df = pd.read_csv(DATA_PATH)
print("=== First 5 rows ===")
print(df.head())
print("=== === === ===  ===")
print("=== Dataset info ===")
df.info()
print("=== === === ===  ===")
print("=== Numeric summary ===")
print(df.describe())
```

```python
duplicates = df[df.duplicated()]
print(duplicates)
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print(f"Rows after removing duplicates: {len(df)}")
```

```python
DISCRETE_THRESHOLD = 10

numeric_cols = df.select_dtypes(include="number").columns.tolist()
discrete_cols = [c for c in numeric_cols if df[c].nunique() <= DISCRETE_THRESHOLD]
continuous_cols = [c for c in numeric_cols if c not in discrete_cols]
categorical_cols = [
    c for c in df.columns
    if c in df.select_dtypes(exclude="number").columns or c in discrete_cols
]

print(f"Continuous  ({len(continuous_cols)}): {continuous_cols}")
print(f"Categorical ({len(categorical_cols)}): {categorical_cols}")
```

```python
def make_grid(n_plots, n_cols=3, cell_size=(5.5, 4)):
    """Create a grid just big enough for n_plots and hide any unused axes."""
    n_cols = min(n_cols, max(n_plots, 1))
    n_rows = math.ceil(n_plots / n_cols)
    fig, axes = plt.subplots(
        n_rows, n_cols,
        figsize=(cell_size[0] * n_cols, cell_size[1] * n_rows),
        squeeze=False,
    )
    axes = axes.flatten()
    for ax in axes[n_plots:]:
        ax.set_visible(False)
    return fig, axes[:n_plots]
```

```python
fig, axes = make_grid(len(continuous_cols))
for ax, col in zip(axes, continuous_cols):
    sns.histplot(df[col], kde=True, bins=30, ax=ax)
    ax.axvline(df[col].mean(), color="red", linestyle="--", label=f"Mean: {df[col].mean():.1f}")
    ax.axvline(df[col].median(), color="green", linestyle="--", label=f"Median: {df[col].median():.1f}")
    ax.set_title(f"Distribution of {col}")
    ax.legend()
plt.tight_layout()
plt.show()

print("Skewness:")
print(df[continuous_cols].skew().round(2))
```
